<h1> Setting Up

```python
#  --- Importing Libraries ---
import pandas as pd

from sqlalchemy import create_engine
```

```python
# --- Load Dataset ---
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
display(df.head(3))

display(df.info())

display(df.describe(include='all'))
```

<h1> Data Preparation

```python
# Convert Headers to Snake Case
df.columns = df.columns.str.lower().str.replace(' ', '_')

# Changing Some Headers' Name
df = df.rename(columns={'purchase_amount_(usd)': 'purchase_amount'})
```

<h2> Handling Null Values

```python
display(df.isnull().sum())
```

```python
# Review Rating: Fill with Median based on Category (to avoid category bias)
df['review_rating'] = df.groupby('category')['review_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
display(df.isnull().sum())
```

<h2> Create New Columns

```python
# Create Column age_group
# display(df[['age']].value_counts().sort_values())

bin_edges = [0, 25, 40, 60, 120]
labels = ['Young Adult', 'Adult', 'Middle-aged', 'Senior']
df['age_group'] = pd.cut(df['age'], bins=bin_edges, labels=labels)

df[['age', 'age_group']].head(10)
```

```python
# Create Column purchase_frequency_days
print(df[['frequency_of_purchases']].value_counts())

frequency_mapping = {
    'Every 3 Months': 90,
    'Annually': 365,
    'Quarterly': 90,
    'Monthly':  30,
    'Bi-Weekly': 14,
    'Fortnightly': 14,
    'Weekly': 7
}

df['purchase_frequency_days'] = df['frequency_of_purchases'].map(frequency_mapping)

df[['frequency_of_purchases', 'purchase_frequency_days']].head(10)
```

<h2> Remove Duplicate Columns

```python
# Checking If discount_applied and promo_code_used Are The Same --> True
(df['discount_applied'] == df['promo_code_used']).all()

# Drop promo_code_used Column
df = df.drop('promo_code_used', axis=1)

df.head(3)
```

<h1> Export To Database

```python
# --- Windows Authentication type MS SQL Database ---
# Connect to Database
server = r'(localdb)\MSSQLLocalDB'
database = 'da_projects'
# username = 'sa'
# password = ''
driver = 'ODBC Driver 18 for SQL Server'

# connection_string = f"mssql+pyodbc://{username}:{password}@{server}/{database}?driver={driver}"
connection_string = rf"mssql+pyodbc://@{server}/{database}?driver={driver}&trusted_connection=yes"
engine = create_engine(connection_string, fast_executemany=True)

# Load Dataframe into Database
table_name = 'customer_behavior'
df.to_sql(
    name=table_name, 
    con=engine, 
    if_exists='replace', 
    index=False,
    schema='dbo'
)

print(f'Data successfully loaded into {table_name} in database {database} ')
```
